```python
import pandas as pd
import numpy as np
```

```python
#busca recolectar y acceder a los datos del archivo que están en formato JSON. Es importante dejar todos los datos de manera accesible, evitando estructuras que recuerden a un archivo JSON, como diccionarios.

#Opcion 2: Desde api

import requests

#Solicitud a la api
response = requests.get(url_franquicias)
#Convirtiendo respuesta de json a objeto python
data = response.json()
#Creando dataframe
datafra= pd.DataFrame(data)
datafra


```

```python
#Continuacion proceso, igual dos opciones

uf = pd.read_json(url_franquicias)
uf
```

```python
#Se tienen dos diccionarios: ventas y tienda

#Normalizacion Ventas
ventas_uf= pd.json_normalize(uf['ventas'])
ventas_uf
```

```python
#Normalizacion tienda

tienda_uf = pd.json_normalize(uf['tienda'])
tienda_uf
```

```python
#Unificacion dataframe

uf_union = pd.concat([uf.drop(['ventas', 'tienda'],axis=1), ventas_uf, tienda_uf], axis=1)
```
