```python
import pandas as pd
import numpy as np
```

```python
#busca recolectar y acceder a los datos del archivo que están en formato JSON. Es importante dejar todos los datos de manera accesible, evitando estructuras que recuerden a un archivo JSON, como diccionarios.

#Opcion 2: Desde api

import requests

#Solicitud a la api
response = requests.get(url_franquicias)
#Convirtiendo respuesta de json a objeto python
data = response.json()
#Creando dataframe
datafra= pd.DataFrame(data)
datafra


```

```python
#Continuacion proceso, igual dos opciones

uf = pd.read_json(url_franquicias)
uf
```

```python
#Se tienen dos diccionarios: ventas y tienda

#Normalizacion Ventas
ventas_uf= pd.json_normalize(uf['ventas'])
ventas_uf
```

```python
#Normalizacion tienda

tienda_uf = pd.json_normalize(uf['tienda'])
tienda_uf
```

```python
#Unificacion dataframe

uf_union = pd.concat([uf.drop(['ventas', 'tienda'],axis=1), ventas_uf, tienda_uf], axis=1)
```

**Limpieza de datos**

```python
#Considerando que la columna ventas_por_dia	tiene 897/898 registros nulos se opta por eliminar

uf_union = uf_union.drop(columns=['ventas_por_dia'])
```

```python
#Se tienen dos valores duplicados por fila

uf_union.duplicated().sum()
```

```python
#Para validar que hacer con estos valores se revisa sus valores

uf_union[uf_union.duplicated()] #Solo muestra la primeras filas de las que estan duplicadas
```

```python
#Validar todas las celdas duplicadas
duplicados = uf_union[uf_union.duplicated(keep=False)]
print(duplicados)
```

```python
#Se opta por eliminar los duplicados

uf_union = uf_union.drop_duplicates()
```

```python
#Comprobacion
uf_union.duplicated().sum()
```

```python
#Ya que se tiene en la columna tamano_tienda	20 registros nulos, se intentara validar los registros para confirmar si se pueden normalizar

uf_union[['tamano_tienda','ID_tienda']]
```

```python
#Muestra las columnas de ID y tamano, donde los valores son nulos
uf_union[uf_union['tamano_tienda'].isnull()][['ID_tienda','tamano_tienda']]
```

```python
#Mediante un sondeo se valida que los ID de estas tiendas son registros unicos
uf_union[uf_union['ID_tienda']== 'OUT642'][['ID_tienda','tamano_tienda']]
```

```python
#Dada esta situacion, se opta por agregar nueva categoria de 'desconocido

uf_union['tamano_tienda'] = uf_union['tamano_tienda'].fillna('desconocido')
```

```python
#Finalmente, la columna cliente_por_dia	tiene 18 registro nulos. Se opta por llenarlos con el promedio

uf_union['cliente_por_dia']= uf_union['cliente_por_dia'].fillna(uf_union['cliente_por_dia'].mean())
```

```python
#Comprobacion

uf_union.isnull().sum()
```

```python
# Con respecto a las diferentes categorias de la columna 'tamano_tienda' (pequeno', 'medio', 'Pequeno')se normaliza con funcion replace

uf_union['tamano_tienda'].replace({'pequeno': 'Pequeño', 'Pequeno':'Pequeño','medio':'Medio', 'desconocido': 'Desconocido'}, inplace=True)
```

```python
#Comprobacion
uf_union['tamano_tienda'].unique()
```

```python
# El siguiente ajuste es en la columna cliente_por_dia que registra como float pero los datos son int. Se cambia formato a int

uf_union['cliente_por_dia'] = uf_union['cliente_por_dia'].astype(int)
```

```python
#Bibliotecas
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Crear grafico de dispersion entre variedad de productos (productos disponibles) y las ventas (ventas_tienda)

#Crear figura y eje
fig, ax = plt.subplots(figsize=(8, 6))
fig.set_facecolor('#ebebeb')
ax.set_facecolor('#ebebeb')

#Graficar el grafico
ax.scatter(uf_union['items_disponibles'], uf_union['ventas_tienda'], color='#e23155', alpha=0.7)

# Otras Configuraciones adicionales
ax.set_xlabel('Productos disponibles')
ax.set_ylabel('Ventas')
ax.set_title('Relacion entre productos disponibles y ventas')
#Se esconde borde superior y derecho
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
#Formato borde izquierdo e inferior
ax.spines['bottom'].set_color("#19325c")  # Color del eje X
ax.spines['left'].set_color("#19325c")
#Cambio formato de las marcas del eje x y y
ax.tick_params(axis='x', colors="#19325c")
ax.tick_params(axis='y', colors="#19325c")

#Evitar superposiciones
plt.tight_layout()
plt.show()
```
