```python
#10 - Un programa debe ser escrito para leer dos números y luego preguntar a la persona usuaria qué operación desea realizar. El resultado de la operación debe incluir información sobre el número, si es par o impar, positivo o negativo, e entero o decimal.
number1 = input('Escribe un numero: ')
number2 = input('Escribe un numero: ')
operacion = input('Indica la operacion que quieres realizar (+ - * /)')
resultado = 0


if "." in number1:
  number1 = float(number1)
else:
  number1=int(number1)

if "." in number2:
  number2 = float(number2)
else:
  number2=int(number2)

if operacion == "+":
  resultado = number1 + number2
  print(resultado)
elif operacion == "-":
  resultado = number1 - number2
  print(resultado)
elif operacion == "*":
  resultado = number1 * number2
  print(resultado)
elif operacion == "/":
  resultado = number1 / number2
  print(resultado)
else:
  print('Operacion no valida')

if resultado%1 == 0:
  print('El resultado es un numero entero')
else:
  print('El resultado es un numero decimal')

if resultado%2 == 0:
  print('El resultado es un numero par')
else:
  print('El resultado es un numero impar')

if resultado>0:
  print('El resultado es un numero positivo')
else:
  print('El resultado es un numero negativo')



```

```python
#11 - Escribe un programa que pida a la persona usuaria tres números que representan los lados de un triángulo. El programa debe informar si los valores pueden utilizarse para formar un triángulo y, en caso afirmativo, si es equilátero, isósceles o escaleno.
lado1 = int(input('Escribe la medida del primer lado: '))
lado2 = int(input('Escribe la medida del segundo lado: '))
lado3 = int(input('Escribe la medida del tercer lado: '))

if lado1+lado2 > lado3 or lado2+lado3 > lado1 or lado1+lado3 > lado2:
  print('Los valores pueden utilizarse para hacer un triangulo del tipo....')
  if lado1 == lado2 == lado3:
    print("Triángulo Equilátero")
  elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
    print('Triángulo Isósceles')
  elif lado1 != lado2 != lado3:
    print('Triángulo Escaleno')
```

```python
#Un establecimiento está vendiendo combustibles con descuentos variables. Para el etanol, si la cantidad comprada es de hasta 15 litros, el descuento será del 2% por litro. En caso contrario, será del 4% por litro. Para el diésel, si la cantidad comprada es de hasta 15 litros, el descuento será del 3% por litro. En caso contrario, será del 5% por litro. El precio por litro de diésel es de R$ 2,00 y el precio por litro de etanol es de R$ 1,70. Escribe un programa que lea la cantidad de litros vendidos y el tipo de combustible (E para etanol y D para diésel) y calcule el valor a pagar por el cliente.
tipo = input('Ingrese el tipo de combustible a comprar (E para etanol y D para diésel): ')
cantidad= float(input('Indique la cantidad de litros comprados: '))
total = 0

if tipo == 'E':
  precio = 1.70
  if cantidad <=15:
    descuento = 0.02
  else:
    descuento = 0.04

if tipo == 'D':
  precio = 2.00
  if cantidad <=15:
    descuento = 0.03
  else:
    descuento = 0.05

total_descuento = precio * (cantidad*descuento)
total = (cantidad *precio) - descuento
print(f'Tipo combustible: {tipo}')
print(f'Precio: {precio}')
print(f'descuento: {total_descuento}')
print(f'Total a pagar: {total}')



```
